```python
9) Eigenvalues and eigenvectors
```

```python
import numpy as np
import matplotlib.pyplot as plt

C = np.array([[3, 1],
              [1, 3]])

values, vectors = np.linalg.eig(C)

print("Eigenvalues:", values)
print("Eigenvectors:")
print(vectors)

for i in range(2):
    plt.quiver(0, 0, vectors[0, i], vectors[1, i],
               angles="xy", scale_units="xy", scale=1)

plt.xlim(-1.5, 1.5)
plt.ylim(-1.5, 1.5)
plt.grid()
plt.title("Eigenvectors")
plt.show()
```

```python
10)Principal direction of sensor data
```

```python
import numpy as np
import matplotlib.pyplot as plt

C = np.array([[5, 2],
              [2, 2]])

values, vectors = np.linalg.eig(C)
i = np.argmax(values)

print("Eigenvalues:", values)
print("Largest Eigenvalue:", values[i])
print("Principal Eigenvector:", vectors[:, i])

for j in range(2):
    plt.quiver(0, 0, vectors[0, j], vectors[1, j],
               angles="xy", scale_units="xy", scale=1)

plt.xlim(-1.2, 1.2)
plt.ylim(-1.2, 1.2)
plt.grid()
plt.title("Principal Directions")
plt.show()
```

```python
12)normal dis
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mean = 70
sd = 10

prob = norm.cdf(80, mean, sd) - norm.cdf(60, mean, sd)
print("Probability:", prob)

x = np.linspace(40, 100, 200)
y = norm.pdf(x, mean, sd)

plt.plot(x, y)
plt.fill_between(x, y, where=(x >= 60) & (x <= 80), alpha=0.4)
plt.title("Normal Distribution")
plt.show()
```

```python
14)Bayes’ theorem for disease detection
```

```python
import matplotlib.pyplot as plt

p_d = 0.02
p_pos_d = 0.95
p_pos_notd = 0.05

posterior = (p_pos_d * p_d) / (
    p_pos_d * p_d + p_pos_notd * (1 - p_d)
)

print("Probability of Disease after Positive Test:", posterior)

plt.bar(["Prior", "Posterior"], [p_d, posterior])
plt.ylabel("Probability")
plt.title("Disease Probability")
plt.show()
```

```python
15: Bayes’ theorem for spam detection
```

```python
import matplotlib.pyplot as plt

p_spam = 0.3
p_prize_spam = 0.7
p_prize_notspam = 0.1

posterior = (p_prize_spam * p_spam) / (
    p_prize_spam * p_spam +
    p_prize_notspam * (1 - p_spam)
)

print("Probability of Spam:", posterior)

plt.bar(["Prior", "Posterior"], [p_spam, posterior])
plt.ylabel("Probability")
plt.title("Spam Probability")
plt.show()
```

```python
18)kl divergence
```

```python
import numpy as np
import matplotlib.pyplot as plt

P = np.array([0.5, 0.3, 0.2])
Q = np.array([0.4, 0.4, 0.2])

kl = np.sum(P * np.log(P / Q))
print("KL Divergence:", kl)

x = np.arange(3)
plt.bar(x - 0.2, P, width=0.4, label="P")
plt.bar(x + 0.2, Q, width=0.4, label="Q")
plt.xticks(x, ["Word 1", "Word 2", "Word 3"])
plt.legend()
plt.show()
```

```python
19)compare entropy values
```

```python
import numpy as np
import matplotlib.pyplot as plt

P = np.array([0.9, 0.1])
Q = np.array([0.5, 0.5])

H1 = -np.sum(P * np.log2(P))
H2 = -np.sum(Q * np.log2(Q))

print("Entropy P:", H1)
print("Entropy Q:", H2)
print("Greater Uncertainty: Q" if H2 > H1 else "Greater Uncertainty: P")

plt.bar(["P", "Q"], [H1, H2])
plt.ylabel("Entropy (bits)")
plt.show()
```

```python
20)Cross-entropy and KL divergence
```

```python
import numpy as np
import matplotlib.pyplot as plt

Y = np.array([0, 1, 0, 0])
P = np.array([0.1, 0.6, 0.2, 0.1])

cross_entropy = -np.sum(Y * np.log(P))
kl = np.sum(Y[Y > 0] * np.log(Y[Y > 0] / P[Y > 0]))

print("Cross-Entropy:", cross_entropy)
print("KL Divergence:", kl)

x = np.arange(4)
plt.bar(x - 0.2, Y, width=0.4, label="Actual")
plt.bar(x + 0.2, P, width=0.4, label="Predicted")
plt.xticks(x, ["Class 1", "Class 2", "Class 3", "Class 4"])
plt.legend()
plt.show()
```

```python
21) Polynomial regression with degrees 1, 2 and 10
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

np.random.seed(1)
X = np.linspace(0, 10, 30)
y = 2 * X**2 + np.random.randn(30) * 10

xx = np.linspace(0, 10, 200)

plt.scatter(X, y, label="Actual")

for d in [1, 2, 10]:
    model = make_pipeline(PolynomialFeatures(d), LinearRegression())
    model.fit(X.reshape(-1, 1), y)
    pred = model.predict(xx.reshape(-1, 1))
    plt.plot(xx, pred, label=f"Degree {d}")

plt.legend()
plt.title("Polynomial Regression")
plt.show()
```

```python
22)Linear regression with nonlinear data
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

np.random.seed(1)
X = np.linspace(1, 10, 100).reshape(-1, 1)
y = (X.ravel() ** 2) + np.random.randn(100) * 5

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1
)

model = LinearRegression()
model.fit(X_train, y_train)

train_pred = model.predict(X_train)
test_pred = model.predict(X_test)

print("Training MSE:", mean_squared_error(y_train, train_pred))
print("Testing MSE:", mean_squared_error(y_test, test_pred))

plt.scatter(X_test, y_test, label="Actual")
plt.scatter(X_test, test_pred, label="Predicted")
plt.legend()
plt.show()
```

```python
23)Compare polynomial degrees 2 and 15
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

np.random.seed(1)
X = np.linspace(0, 6, 40).reshape(-1, 1)
y = np.sin(X.ravel()) + np.random.randn(40) * 0.1

X_train, X_test = X[:25], X[25:]
y_train, y_test = y[:25], y[25:]

xx = np.linspace(0, 6, 200).reshape(-1, 1)

plt.scatter(X, y, label="Actual")

for d in [2, 15]:
    model = make_pipeline(PolynomialFeatures(d), LinearRegression())
    model.fit(X_train, y_train)

    print("Degree:", d)
    print("Training MSE:", mean_squared_error(y_train, model.predict(X_train)))
    print("Testing MSE:", mean_squared_error(y_test, model.predict(X_test)))

    plt.plot(xx, model.predict(xx), label=f"Degree {d}")

plt.ylim(-3, 3)
plt.legend()
plt.show()
```

```python
24: Compare temperature prediction models
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

np.random.seed(1)
X = np.linspace(0, 10, 100).reshape(-1, 1)
y = 25 + 5 * np.sin(X.ravel()) + np.random.randn(100)

X_train, X_test = X[:70], X[70:]
y_train, y_test = y[:70], y[70:]

xx = np.linspace(0, 10, 200).reshape(-1, 1)

plt.scatter(X, y, label="Actual")

for d in [1, 3, 12]:
    model = make_pipeline(PolynomialFeatures(d), LinearRegression())
    model.fit(X_train, y_train)

    train_error = mean_squared_error(y_train, model.predict(X_train))
    test_error = mean_squared_error(y_test, model.predict(X_test))

    print("Degree:", d, "Training MSE:", train_error,
          "Testing MSE:", test_error)

    plt.plot(xx, model.predict(xx), label=f"Degree {d}")

plt.legend()
plt.title("Temperature Prediction")
plt.show()
```

```python
25: Learning curves for linear and polynomial regression
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

np.random.seed(1)
X = np.linspace(0, 10, 100).reshape(-1, 1)
y = X.ravel()**2 + np.random.randn(100) * 5

sizes = [10, 20, 30, 40, 50, 60, 70]
models = {
    "Linear": LinearRegression(),
    "Polynomial": make_pipeline(
        PolynomialFeatures(10), LinearRegression()
    )
}

for name, model in models.items():
    train_errors = []
    val_errors = []

    for n in sizes:
        model.fit(X[:n], y[:n])
        train_errors.append(mean_squared_error(y[:n], model.predict(X[:n])))
        val_errors.append(mean_squared_error(y[70:], model.predict(X[70:])))

    plt.plot(sizes, train_errors, marker="o", label=name + " Training")
    plt.plot(sizes, val_errors, marker="s", label=name + " Validation")

plt.xlabel("Training Samples")
plt.ylabel("Mean Squared Error")
plt.title("Learning Curves")
plt.legend()
plt.show()
```
