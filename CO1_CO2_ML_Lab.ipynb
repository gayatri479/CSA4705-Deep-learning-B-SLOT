## Question 3
**Determinant, inverse and rank for an encryption matrix**

```python
import numpy as np
M = np.array([[3, 3], [2, 5]], dtype=float)
det = np.linalg.det(M)
rank = np.linalg.matrix_rank(M)
print("Determinant:", round(det, 3))
print("Rank:", rank)
if not np.isclose(det, 0):
    print("Inverse:\n", np.linalg.inv(M))
    print("Suitable for square-matrix encryption: YES (invertible)")
else:
    print("Suitable for reversible matrix encryption: NO (singular)")
```

## Question 4
**Page ranking using eigenvalues and eigenvectors**

```python
import numpy as np
# Column j gives the probability of moving from page j to each page.
P = np.array([[0, 0.5, 1/3],
              [0.5, 0, 1/3],
              [0.5, 0.5, 1/3]], dtype=float)
values, vectors = np.linalg.eig(P)
i = np.argmin(np.abs(values - 1))
rank_scores = np.abs(np.real(vectors[:, i]))
rank_scores = rank_scores / rank_scores.sum()
print("Transition matrix:\n", P)
print("Eigenvalues:", np.round(values, 4))
print("Page importance scores:", np.round(rank_scores, 4))
print("Most important page: Page", np.argmax(rank_scores) + 1)
```

```python
import numpy as np
customer1 = np.array([5, 2, 0, 1])
customer2 = np.array([4, 1, 0, 2])
customer3 = np.array([0, 1, 5, 4])
def cosine(a, b):
    denom = np.linalg.norm(a) * np.linalg.norm(b)
    return float(np.dot(a, b) / denom) if denom else 0.0
for name, c in [("Customer 2", customer2), ("Customer 3", customer3)]:
    print(name, "dot product:", np.dot(customer1, c),
          "cosine similarity:", round(cosine(customer1, c), 3))
```

## Question 15
**Batch, stochastic and mini-batch gradient descent**

```python
import numpy as np
np.random.seed(1)
X = np.linspace(0, 10, 100)
y = 2*X + 1 + np.random.normal(0, 1, 100)
def train_gd(mode, epochs=30, lr=0.01, batch_size=10):
    w, b, losses = 0.0, 0.0, []
    for _ in range(epochs):
        if mode == "batch":
            batches = [np.arange(len(X))]
        elif mode == "stochastic":
            batches = [np.array([i]) for i in np.random.permutation(len(X))]
        else:
            order = np.random.permutation(len(X))
            batches = [order[i:i+batch_size] for i in range(0, len(X), batch_size)]
        for idx in batches:
            err = (w*X[idx] + b) - y[idx]
            dw = 2*np.mean(err*X[idx]); db = 2*np.mean(err)
            w -= lr*dw; b -= lr*db
        losses.append(np.mean((w*X+b-y)**2))
    return w, b, losses
for mode in ["batch", "stochastic", "mini-batch"]:
    w, b, losses = train_gd(mode)
    print(mode, "w=", round(w,3), "b=", round(b,3), "final MSE=", round(losses[-1],3))
```

## Question 22
**Grid search, random search and Bayesian-style optimization**

```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
X,y = load_breast_cancer(return_X_y=True)
Xtr,Xte,ytr,yte = train_test_split(X,y,test_size=0.25,stratify=y,random_state=42)
base = RandomForestClassifier(random_state=42)
grid_params = {"n_estimators":[50,100],"max_depth":[None,5,10]}
grid = GridSearchCV(base,grid_params,cv=3,n_jobs=-1).fit(Xtr,ytr)
random = RandomizedSearchCV(base,{"n_estimators":[50,100,150,200],
    "max_depth":[None,3,5,8,12],"min_samples_split":[2,4,6]},
    n_iter=8,cv=3,random_state=42,n_jobs=-1).fit(Xtr,ytr)
print("Grid best:",grid.best_params_,"test accuracy:",accuracy_score(yte,grid.predict(Xte)))
print("Random best:",random.best_params_,"test accuracy:",accuracy_score(yte,random.predict(Xte)))
# Simple sequential/model-based-inspired search (not a full Gaussian-process BO library)
candidates = [3,5,7,9,12,15]
scores = []
for depth in candidates:
    m = RandomForestClassifier(n_estimators=100,max_depth=depth,random_state=42)
    m.fit(Xtr,ytr); scores.append(m.score(Xte,yte))
best = int(np.argmax(scores))
print("Sequential search best depth:",candidates[best],"test accuracy:",scores[best])
print("For formal Bayesian optimization, install and use Optuna or scikit-optimize.")
```

## Question 23
**Reinforcement-learning agent for traffic signals (Q-learning)**

```python
import numpy as np
np.random.seed(0)
Q = np.zeros((3,2))
alpha, gamma, epsilon = 0.2, 0.9, 0.15
def next_state_reward(state, action):
    if action == 0:
        reward = [2, 1, -2][state]
        next_state = max(0,state-1) if state>0 else 0
    else:
        reward = [1, 2, 3][state]
        next_state = max(0,state-1)
    return next_state, reward
for episode in range(1000):
    state = np.random.randint(0,3)
    for _ in range(20):
        action = np.random.randint(2) if np.random.rand()<epsilon else np.argmax(Q[state])
        ns,reward = next_state_reward(state,action)
        Q[state,action] += alpha*(reward + gamma*np.max(Q[ns]) - Q[state,action])
        state = ns
print("Learned Q table:\n",np.round(Q,2))
for s in range(3):
    print(["Low","Medium","High"][s],"traffic -> action",np.argmax(Q[s]))
```

## Question 30
**Backpropagation: calculate gradients and update weights**

```python
import numpy as np
sigmoid = lambda z: 1/(1+np.exp(-z))
x = np.array([[0.5],[0.8]])
y = np.array([[1.0]])
W1 = np.array([[0.2,0.4],[0.3,0.1]])
b1 = np.zeros((2,1))
W2 = np.array([[0.5,0.6]])
b2 = np.zeros((1,1))
lr = 0.1

z1 = W1@x+b1; a1 = sigmoid(z1)
z2 = W2@a1+b2; a2 = sigmoid(z2)
dz2 = a2-y
dW2 = dz2@a1.T; db2 = dz2
dz1 = (W2.T@dz2)*a1*(1-a1)
dW1 = dz1@x.T; db1 = dz1
print("Prediction before update:",a2.item())
print("Output weight gradient:\n",dW2)
W2 -= lr*dW2; b2 -= lr*db2
W1 -= lr*dW1; b1 -= lr*db1
a1_new = sigmoid(W1@x+b1)
a2_new = sigmoid(W2@a1_new+b2)
print("Prediction after update:",a2_new.item())
```

## Question 32
**Binary cross-entropy for fraud detection**

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import make_pipeline
from sklearn.metrics import log_loss, classification_report, confusion_matrix, roc_auc_score
X,y = make_classification(n_samples=1000,n_features=12,n_informative=7,
    weights=[0.9,0.1],flip_y=0.02,random_state=42)
Xtr,Xte,ytr,yte = train_test_split(X,y,test_size=0.25,stratify=y,random_state=42)
model = make_pipeline(StandardScaler(),
    MLPClassifier(hidden_layer_sizes=(32,16),max_iter=500,
                  early_stopping=True,random_state=42))
model.fit(Xtr,ytr)
prob = model.predict_proba(Xte)[:,1]
pred = (prob >= 0.5).astype(int)
print("Binary cross-entropy:",log_loss(yte,prob))
print("ROC-AUC:",roc_auc_score(yte,prob))
print("Confusion matrix:\n",confusion_matrix(yte,pred))
print(classification_report(yte,pred,target_names=["Legitimate","Fraud"]))
```
